El código debe recopilar los datos de cantidad de ventas durante los años 2022 y 2023 y calcular la variación porcentual. A partir del valor de la variación, se deben proporcionar las siguientes sugerencias:

Para una variación superior al 20%: bonificación para el equipo de ventas.
Para una variación entre el 2% y el 20%: pequeña bonificación para el equipo de ventas.
Para una variación entre el 2% y el -10%: planificación de políticas de incentivo a las ventas.
Para bonificaciones inferiores al -10%: recorte de gastos.

```python
# Recolectamos las ventas de los dos años
venta_2022 = float(input('Ingrese la cantidad de ventas en 2022: '))
venta_2023 = float(input('Ingrese la cantidad de ventas en 2023: '))

# Calculamos la variación porcentual entre las ventas de los años 2022 y 2023
var_porcentual = 100 * (venta_2023 - venta_2022) / (venta_2022)

# Análisis condicional de la variación porcentual para determinar la sugerencia a enviar
if var_porcentual > 20:
    print('Bonificación para el equipo de ventas.')
elif 2 <= var_porcentual <= 20:
    print('Pequeña bonificación para el equipo de ventas.')
elif -10 <= var_porcentual < 2:
    print('Planificación de políticas de incentivo a las ventas.')
else:
    print('Recorte de gastos.')
```

Cuando trabajamos con bucles, podemos controlar el flujo de ejecución dentro del bloque de código, lo que nos permite manipular la ejecución de los bucles. continue y break son los comandos de control que podemos usar con los bucles for y while.

continue interrumpe la iteración actual del bucle y salta a la siguiente, es decir, regresa al inicio del código. Como ejemplo, aquí hay un código que cuenta del 1 al 5 con un bucle for:


Este código imprime todos los números del 1 al 5, excepto el 4. Cuando el valor de i es 4, continue salta a la siguiente iteración, omitiendo la instrucción print después de la condición en la iteración actual.

```python
for i in range(1, 6):
    if i == 4:
        continue
    print(i)
```

Por otro lado, break detiene por completo la ejecución del bucle, saliendo del bloque de código. Utilicemos el mismo ejemplo de conteo, pero esta vez con break:

```python
for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

Crea un programa que tenga la siguiente lista con los gastos de una empresa de papel [2172.54, 3701.35, 3518.09, 3456.61, 3249.38, 2840.82, 3891.45, 3075.26, 2317.64, 3219.08]. Con estos valores, crea un programa que calcule el promedio de gastos. Sugerencia: usa las funciones integradas sum() y len().

```python
gastos = [2172.54, 3701.35, 3518.09, 3456.61, 3249.38, 2840.82, 3891.45, 3075.26, 2317.64, 3219.08]

total_gastos = sum(gastos)
numero_gastos = len(gastos)
promedio_gastos = total_gastos / numero_gastos

print(f'El total de gastos es: {total_gastos:.2f}')
print(f'El número de gastos es: {numero_gastos}')
print(f'El promedio de gastos es: {promedio_gastos:.2f}')
```

```python
# Datos de gastos
gastos = [2172.54, 3701.35, 3518.09, 3456.61, 3249.38, 2840.82, 3891.45, 3075.26, 2317.64, 3219.08]

# Calculamos el promedio encontrando el valor total de gastos con la función sum
# y la cantidad total de compras realizadas con len
total_gastos = sum(gastos)
cantidad_compras = len(gastos)
media_gastos = total_gastos / cantidad_compras
# Resultado
print(f'El promedio de gastos es {media_gastos} reales.')
```

Una empresa evalúa su trimestre con base en:

Ingresos totales
Gastos totales
Número de nuevos clientes
Clasificación:

Si ingresos - gastos > $10,000 y más de 50 nuevos clientes → "Trimestre Excelente"
Si ingresos - gastos > $5,000 y al menos 20 clientes → "Trimestre Bueno"
Si ingresos - gastos > 0 → "Trimestre Regular"
Si ingresos - gastos ≤ 0 → "Trimestre Deficitario"

```python
# Solicitar datos del trimestre
ingresos = float(input('Ingrese los ingresos totales: '))
gastos = float(input('Ingrese los gastos totales: '))
nuevos_clientes = int(input('Ingrese el número de nuevos clientes: '))

# Calcular el saldo
saldo = ingresos - gastos

# Clasificar el trimestre según las reglas
if saldo > 10000 and nuevos_clientes > 50:
    print('Trimestre Excelente')
elif saldo > 5000 and nuevos_clientes >= 20:
    print('Trimestre Bueno')
elif saldo > 0:
    print('Trimestre Regular')
else:
    print('Trimestre Deficitario')
```

Ana está desarrollando un programa que necesita procesar una lista de 5 nombres de clientes para generar informes mensuales. Para ello, necesita escribir un programa que recorra la lista de nombres y muestre cada cliente.

clientes = ["Juan", "Maria", "Carlos", "Ana", "Beatriz"]

Ayuda a Ana a decidir entre usar un lazo for o while. Escribe el programa usando el lazo que creas más adecuado para esta tarea y explica por qué elegiste ese lazo.

### Elección del Lazo: for

He elegido el lazo **`for`** porque en Python es la forma más 'pythónica' y eficiente de recorrer una colección de elementos (como una lista).

**Ventajas:**
1. **Simplicidad:** No necesitas gestionar manualmente un índice o contador.
2. **Legibilidad:** El código es más fácil de leer y entender ('para cada cliente en la lista...').
3. **Seguridad:** Elimina el riesgo de bucles infinitos por olvidar actualizar la condición de parada.

```python
clientes = ["Juan", "Maria", "Carlos", "Ana", "Beatriz"]

print("Informe Mensual de Clientes:")
for cliente in clientes:
    print(f"Cliente: {cliente}")
```

André está probando una nueva función en el backend de Buscante que procesa datos en un bucle. Durante las pruebas, se dio cuenta de que el sistema dejó de responder y sospecha que el problema está en un bucle infinito.

contador = 0

while contador < 10:
    print("Procesando datos...")

    ¿Cuál es el problema del código de André y cómo resolverlo?

### Análisis del Problema

El código de André tiene un **bucle infinito** porque la condición de salida depende de la variable `contador`, pero esta nunca se actualiza dentro del bloque `while`.

**Solución:**
Se debe añadir una línea de código que incremente el valor de `contador` (por ejemplo, `contador += 1`) en cada vuelta del ciclo. De esta forma, eventualmente llegará a 10 y la condición `contador < 10` será falsa, permitiendo que el programa termine.

```python
contador = 0

# Versión corregida:
while contador < 10:
    print(f"Procesando datos... (Iteración {contador + 1})")
    # Incrementamos el contador para evitar el bucle infinito
    contador += 1

print("Procesamiento finalizado.")
```

```python
contador = 0

while contador < 10:
    print("Procesando datos...")
    contador += 1  # Actualiza el contador para evitar el bucle infinito
```

João está desarrollando un sistema de registro para un sitio de lectura. Necesita asegurarse de que los usuarios ingresen un nombre de usuario y una contraseña válidos. Las reglas son las siguientes:

El nombre de usuario debe tener al menos 5 caracteres.
La contraseña debe tener al menos 8 caracteres.
João quiere que el sistema siga solicitando la información hasta que ambas condiciones se cumplan. Cuando el usuario ingresa datos válidos, el programa debe mostrar el mensaje: "¡Registro realizado con éxito!".

Crea un programa que implemente esta lógica usando un bucle while.

```python
while True:
    usuario = input('Ingrese su nombre de usuario (mín. 5 caracteres): ')
    password = input('Ingrese su contraseña (mín. 8 caracteres): ')

    if len(usuario) >= 5 and len(password) >= 8:
        print('¡Registro realizado con éxito!')
        break
    else:
        print('Error: El usuario debe tener al menos 5 caracteres y la contraseña al menos 8. Inténtelo de nuevo.\n')
```
